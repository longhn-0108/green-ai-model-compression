# Green AI experiment visualization

Loads the tracked results from `../results/experiment_log.csv` and plots accuracy, runtime, and estimated energy. Historical results are treated as observations rather than statistical estimates.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../results/experiment_log.csv')
df.head()
```

```python
gpu = df[df['total_time_min'] < 200].copy()
gpu['label'] = gpu.apply(lambda r: 'MobileNetV2' if r.model_name == 'mobilenet_v2' else ('ResNet-50 baseline' if r.technique == 'baseline' else f"ResNet-50 prune {int(r.pruning_amount*100)}%"), axis=1)
ax = gpu.plot.bar(x='label', y='test_accuracy', legend=False, figsize=(10,5), title='Test accuracy — logged GPU runs')
ax.set_ylabel('Accuracy')
plt.xticks(rotation=20)
plt.tight_layout()
```

```python
ax = gpu.plot.bar(x='label', y='total_energy_kwh', legend=False, figsize=(10,5), title='Estimated energy — logged GPU runs')
ax.set_ylabel('Energy (kWh)')
plt.xticks(rotation=20)
plt.tight_layout()
```
